numpy array and basics

```python
import numpy as np
```

array creation

```python
#1d array creations
arr_1d=np.array([10,20,30,40])
print(arr_1d,type(arr_1d))
```

```python
#2d array creations
arr_2d=np.array([[10,20,30],[40,50,60]])
print(arr_2d)
arr_2d=np.array([[10,20,30],
                 [40,50,60]])
print(arr_2d)
```

```python
arr_2d=np.array([[10,20,30.45],
                 [40,50,60]])
print(arr_2d)
```
